Question 3:

```r
# Question 3: Adding 'fill' mapped to the 'cut' factor variable
ggplot(diamonds) +
  layer(
    geom = "bar",
    stat = "bin",
    position = "stack",
    mapping = aes(x = carat, fill = cut),
    data = NULL,
    params = list(bins = 30)
  )
```

**Question 4 answer**: The default position adjustment used in a standard histogram is "stack" (which places the groups on top of each other).

Changing it to "dodge" places the bars for each cut category side-by-side rather than stacked.

```r
# Question 4: Changing position adjustment to "dodge"
ggplot(diamonds) +
  layer(
    geom = "bar",
    stat = "bin",
    position = "dodge",
    mapping = aes(x = carat, fill = cut),
    data = NULL,
    params = list(bins = 30)
  )
```
